```python
import torch
import torch.nn as nn
from matplotlib import pyplot as plt

```

```python
torch.sort(torch.randn(100) * 10)
```

```python
# 평균 0, 표준편차 1인 100개의 데이터에 10을 곱해 표준편차를 10으로 바꾸고, values만 가져온다
x = torch.sort(torch.randn(100) * 10)[0] 
x
```

# Sigmoid

```python
act = nn.Sigmoid() # 각 입력값을 0~1사이 확률값으로 매핑 (객체 지향 방식)
print(act(x))
```

```python
torch.sigmoid(x) # 함수형 호출 방식
```

```python
# matplolib 시각화시 Tensor
plt.plot(x.numpy())
plt.show()
```

# TanH

```python
act = nn.Tanh() # 객체지향 방식
print(act(x)) 
```

```python
torch.tanh(x) # 함수 호출 방식
```

```python
plt.plot(torch.tanh(x).numpy())
plt.show()
```

ReLU

```python
act = nn.ReLU() # 객체 지향 방식
act(x)
```

```python
torch.relu(x) # 함수 호출 방식
```

Leaky ReLU

```python
act = nn.LeakyReLU(negative_slope=0.1) # 객체 지향 방식
act(x)
```

```python
import torch.nn.functional as F
F.leaky_relu(x, negative_slope=0.1)
```

```python
plt.plot(x.numpy(), act(x).numpy())
plt.show()
```

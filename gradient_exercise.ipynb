```python
def numerical_diff(f, x, h=1e-4):
    return (f(x + h) - f(x - h)) / (2 * h)

def function(x):
    return x**2

print(numerical_diff(function, 3.0))
```

수치 미분은 조금 움직였을때 값이 얼마나 변하는지 기울기를 추정할때 사용한다.

딥러닝에서는 실제 학습에서 사용하지는 않음

```python
# 다변화 함수에서 특정 변수에 대한 편미분 값을 계산하는 함수
def partial_diff(f, x, y, var='x', h=1e-4):
    if var == 'x':
        return (f(x + h, y) - f(x - h, y)) / (2 * h)
    else:
        return (f(x, y + h) - f(x, y - h)) / (2 * h)

def function(x, y):
    return x**2 + y**2

print(partial_diff(function, 3.0, 2.0, 'x'))
print(partial_diff(function, 3.0, 2.0, 'y'))
```

편미분은 다변화 함수에서 특정 변수 하나가 결과에 미치는 영향을 수식으로 나타낸 것  

자동 미분은
1. 모든 연산을 계산 그래프로 표현
2. 각 연산 노드에서 국소적인 미분 계산
3. 연쇄 법칙을 사용해 각 가중치에 대한 편미분 값을 계산

결론적으로 자동미분은 그 편미분 값을 계산 그래프와 연쇄 법칙을 이용해서 컴퓨터가 자동으로 계산해주는 기법

```python
def gradient_descent(f, init_x, lr=0.1, step_num=100):
    x = init_x
    for _ in range(step_num):
        grad = numerical_diff(f, x)
        x -= lr * grad
    return x

def function(x):
    return x**2

print(gradient_descent(function, 10.0))
```

손실을 최소화하기 위해서 손실의 반대방향으로 가중치를 수정한다.

$w \leftarrow w - \eta \frac{\partial L}{\partial w}$

w : 현재 가중치  
<- 새 값으로 업데이트 (:=)  
n (eta) 학습률  
∂L / ∂w : 가중치 w가 손실함수 L에 미치는 영향 (편미분, 기울기)  
\- 는 손실을 줄이기 위해 기울기의 반대방향으로 이동

결론적으로 현재의 가중치에서 손실을 가장 빠르게 증가시키는 방향의 반대 방향으로 학습률만큼 이동하여 가중치 갱신(손실을 줄이는 효율적인 이동 방향)

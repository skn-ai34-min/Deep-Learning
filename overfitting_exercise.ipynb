```python
import torch
import torch.nn as nn
import torch.optim as optim

# ex) 이미지 분류할때 보통 사용
class NeuralNet(nn.Module):
    def __init__(self):
        super(NeuralNet, self).__init__() # 부모 클래스 nn.Moudle 초기화
        self.fc1 = nn.Linear(784, 128) # 입력(784) -> 은닉1(128) 선형 변환
        self.bn1 = nn.BatchNorm1d(128) # 은닉1 출력(128차원) 배치 정규화

        self.fc2 = nn.Linear(128, 64)  # 은닉2
        self.bn2 = nn.BatchNorm1d(64)  # 은닉2 출력(64차원) 배치 정규화

        self.fc3 = nn.Linear(64, 10)   # 입력(64) -> 출력(10 클래스) 선형 변환
    
    def forward(self, x):
        x = torch.relu(self.bn1(self.fc1(x)))
        x = torch.relu(self.bn2(self.fc2(x)))
        x = self.fc3(x)
        return x

model = NeuralNet()
model
```

```python
class NeuralNet(nn.Module):
    def __init__(self):
        super(NeuralNet, self).__init__()
        self.fc1 = nn.Linear(784, 128)
        self.dropout1 = nn.Dropout(0.5)
        self.fc2 = nn.Linear(128, 64)
        self.dropout2 = nn.Dropout(0.3)
        self.fc3 = nn.Linear(64, 10)
    
    def forward(self, x):
        x = torch.relu(self.dropout1(self.fc1(x)))
        x = torch.relu(self.dropout2(self.fc2(x)))
        x = self.fc3(x)
        return x

model = NeuralNet()
model
```

만약에 BatchNormalization과 Dropout을 함께 쓴다면 Linear -> BN -> ReLU -> Dropout 처럼 Dropout을 뒤에 배치 (권장)  
서로 붙여서 사용하면 오히려 학습에 방해되는 경우가 생길 수 있음

```python
from skorch import NeuralNetClassifier
from sklearn.model_selection import RandomizedSearchCV
from scipy.stats import uniform

net = NeuralNetClassifier(NeuralNet, max_epochs=10, lr=0.01)
param_dist = { 'lr': uniform(0.0001, 0.1) }
random_search = RandomizedSearchCV(estimator=net, param_distributions=param_dist, n_iter=5, cv=3, verbose=1)
```
